```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from scipy import stats
```

```python
df = pd.read_csv('WineQT.csv')
if 'Id' in df.columns:
    df = df.drop(columns=['Id'])
df.head(3)
```

```python
print(df.describe())
```

```python
sns.pairplot(df, vars=['alcohol', 'volatile acidity', 'citric acid', 'quality'], kind="reg")
plt.show()
```

```python
sns.set(style="ticks", color_codes=True)
sns.pairplot(df, height=2, vars=['fixed acidity', 'volatile acidity', 'alcohol'], hue="quality")
plt.show()
```

```python
corr = stats.pearsonr(df["quality"], df["alcohol"])
print("p-value:\t", corr[1])
print("cor:\t\t", corr[0])
```

```python
corr_va = stats.pearsonr(df["quality"], df["volatile acidity"])
print("p-value:\t", corr_va[1])
print("cor:\t\t", corr_va[0])
```

```python
correlation = df.corr(method='pearson', numeric_only=True)
correlation
```

```python
sns.heatmap(correlation, xticklabels=correlation.columns, yticklabels=correlation.columns)
plt.show()
```
