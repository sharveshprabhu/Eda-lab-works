```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Iris.csv')
if 'Id' in df.columns:
    df = df.drop(columns=['Id'])
df.head()
```

```python
print(df.info())
print(df.describe())
```

```python
df['Species'].value_counts().plot(kind='bar', color=['skyblue', 'salmon', 'lightgreen'], figsize=(8, 5))
plt.title('Distribution of Iris Species')
plt.xlabel('Species')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

```python
df.hist(figsize=(10, 8), bins=15, edgecolor='black', color='teal')
plt.suptitle('Distribution of Floral Measurements')
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
sns.boxplot(x='Species', y='SepalLengthCm', data=df, ax=axes[0, 0])
sns.boxplot(x='Species', y='SepalWidthCm', data=df, ax=axes[0, 1])
sns.boxplot(x='Species', y='PetalLengthCm', data=df, ax=axes[1, 0])
sns.boxplot(x='Species', y='PetalWidthCm', data=df, ax=axes[1, 1])
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(x='PetalLengthCm', y='PetalWidthCm', hue='Species', data=df, s=70)
plt.title('Petal Length vs Petal Width by Species')
plt.show()
```

```python
sns.pairplot(df, hue='Species', markers=['o', 's', 'D'])
plt.show()
```

```python
numeric_df = df.drop(columns=['Species'])
correlation = numeric_df.corr()
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Iris Measurements')
plt.show()
```
