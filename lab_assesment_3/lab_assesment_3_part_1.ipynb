```python
import numpy as np
import pandas as pd
```

```python
height = [120, 122, 125, 127, 121, 123, 137, 131, 161, 145, 141, 132]
bins = [118, 125, 135, 160, 200]
category = pd.cut(height, bins)
category
```

```python
pd.Series(category).value_counts()
```

```python
bin_names = ['Short Height', 'Average height', 'Good Height', 'Taller']
pd.cut(height, bins, labels=bin_names)
```

```python
pd.cut(np.random.rand(40), 5, precision=2)
```

```python
randomNumbers = np.random.rand(2000)
category3 = pd.qcut(randomNumbers, 4)
category3
```

```python
pd.Series(category3).value_counts()
```

```python
df = pd.read_csv('CreditCard/fraudTest.csv')
df.head()
```

```python
amt_bins = [0, 25, 100, 500, df['amt'].max()]
amt_labels = ['Low Spent', 'Medium Spent', 'High Spent', 'Very High Spent']
df['amt_binned'] = pd.cut(df['amt'], bins=amt_bins, labels=amt_labels)
df['amt_binned'].value_counts()
```

```python
df['amt_quartiles'] = pd.qcut(df['amt'], q=4)
df['amt_quartiles'].value_counts()
```

```python
pop_bins = [0, 5000, 50000, 500000, df['city_pop'].max()]
pop_labels = ['Rural', 'Town', 'City', 'Metro']
df['pop_category'] = pd.cut(df['city_pop'], bins=pop_bins, labels=pop_labels)
df['pop_category'].value_counts()
```
