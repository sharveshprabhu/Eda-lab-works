```python
import numpy as np
import pandas as pd
```

```python
df_credit = pd.read_csv('CreditCard/fraudTest.csv')
df_credit['amt'].describe()
```

```python
TotalTransaction = df_credit['amt']
TotalTransaction[np.abs(TotalTransaction) > 1000]
```

```python
df_credit[np.abs(TotalTransaction) > 2000]
```

```python
q25 = df_credit['amt'].quantile(0.25)
q75 = df_credit['amt'].quantile(0.75)
iqr = q75 - q25
lower_bound = q25 - 1.5 * iqr
upper_bound = q75 + 1.5 * iqr
outliers_iqr = df_credit[(df_credit['amt'] < lower_bound) | (df_credit['amt'] > upper_bound)]
print("Number of outliers (IQR):", len(outliers_iqr))
outliers_iqr.head()
```

```python
z_scores = (df_credit['amt'] - df_credit['amt'].mean()) / df_credit['amt'].std()
outliers_z = df_credit[np.abs(z_scores) > 3]
print("Number of outliers (Z-score > 3):", len(outliers_z))
outliers_z.head()
```
